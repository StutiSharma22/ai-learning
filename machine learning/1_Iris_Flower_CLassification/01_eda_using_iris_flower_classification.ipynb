# **Project Overview**

**Project title**- Exploratory data analysis using Iris Flower Classification dataset.


**Problem statement**- The Iris flower classification problem is a supervised machine learning task which aims to identify species based on their respective sepal lengths and petal lengths.

Three-species-of-IRIS-flower.webp

**Project objective**- The objective is to make a predictive model that can classify the iris flower species into its three kinds- Iris-Versicolor, Iris-Virginica and Iris-Setosa.

**Purpose of EDA**- The primary to purpose of the eda is understand data and its underlying structure before building models or applying any techniques.

**Dataset source**- https://www.kaggle.com/datasets/uciml/iris

**Dataset description**- The dataset contains 150 instances. Each instance is a plant. This is one of the earliest datasets used in the literature on classification methods and widely used in statistics and machine learning.  The data set contains 3 classes of 50 instances each, where each class refers to a type of iris plant.  One class is linearly separable from the other 2; the latter are not linearly separable from each other.
The columns dataset contains are-
*   SepalLengthCm- Length of the plant sepal in centimeters.
*   SepalWidthCm- Width of the plant sepal in centimeters.
*   PetalLengthCm- Length of the plant petal in centimeters.
*   PetalWidthCm- Width of the plant petal in centimeters.
*   Species- Name of the iris plant species to which the plant belongs.





**Expected outcome**- A well defined exploratory data analysis that helps to predict the species of the iris plant with a high accuracy.

**Potential Outlier Detection**

```python
for col in features.columns:
  Q1= data[col].quantile(0.25)
  Q3= data[col].quantile(0.75)
  IQR= Q3 - Q1

  lower_bound= Q1 - 1.5 * IQR
  upper_bound= Q3 + 1.5 * IQR

  outliers= data[
      (data[col] < lower_bound) |
      (data[col] > upper_bound)
      ]

  print(f"Feature: {col}")
  print(f"Lower Bound: {lower_bound:.2f}")
  print(f"Upper Bound: {upper_bound:.2f}")
  print(f"Number of potential outliers: {len(outliers)}")
  print(outliers[col].to_list())


```

```python
n_cols= 2
n_rows= math.ceil(len(features.columns) / n_cols)
fig, axes= plt.subplots(n_rows, n_cols, figsize=(12, 4 * n_rows))
axes= axes.flatten()
for i, col in enumerate(features.columns):
  ax= axes[i]
  sns.boxplot(data= data[col], ax=ax)
  ax.set_xlabel(f"Feature: {col}")
  ax.set_ylabel("Values")
fig.suptitle("Box plot for potential outliers")
plt.tight_layout()
plt.show()
```

Observations
- There are no significant outliers detected.
- Investigation for the 4 outliers by species is required and examination whether they represent unusual but plausible observations.
